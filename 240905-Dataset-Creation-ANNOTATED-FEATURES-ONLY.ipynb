```python
import pandas as pd
```

```python
# Load the CSV file
csv_file = '240617-NTA-Norm-Paper2-RAWVAL.csv'
df_csv = pd.read_csv(csv_file)

# Load the XLSX file
xlsx_file = '240715-NTA-Paper2-KARL-RFE-Round1-AnnotatedList.xlsx'
df_xlsx = pd.read_excel(xlsx_file, sheet_name=0)  # Assuming all data is in the first sheet

# Create a list to hold the features we want to keep
features_to_keep = []

# Loop through the pairs of columns in the XLSX file
for i in range(0, df_xlsx.shape[1], 2):
    feature_col = df_xlsx.columns[i]
    match_col = df_xlsx.columns[i + 1]

    # Filter out features where the match is 'No Match'
    valid_features = df_xlsx[df_xlsx[match_col] != 'No Match'][feature_col].tolist()

    # Add the valid features to the list
    features_to_keep.extend(valid_features)

# Find the intersection of the features in the CSV and those to keep from the XLSX
features_in_csv = df_csv.columns.tolist()[1:]  # Skip the 'Type' column
filtered_features = ['Type'] + [feature for feature in features_in_csv if feature in features_to_keep]

# Filter the CSV to only retain the selected features
df_filtered = df_csv[filtered_features]

# Save the filtered CSV to a new file
output_file = '240905-NTA-Paper2-Only-Annotated-RAWVAL.csv'
df_filtered.to_csv(output_file, index=False)

print(f"Filtered CSV saved as {output_file}")
```
