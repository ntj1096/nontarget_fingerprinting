```python
#For normalized data
import pandas as pd
```

### For remaining unannotated features

```python
#Read in important features file
df1_imp = pd.read_csv(r'240702-NTA-Paper2-RFE-Unknown-Features.csv', header=0)
print(df1_imp.shape)

#Read in normalized master to conduct WIFF file matching
df2_wif = pd.read_csv(r'240702-NTA-Paper2-Full-Dataset-WIFFKEY.csv', header=0)
print(df2_wif.shape)
```

```python
# Create a new DataFrame to store the results
result_df = pd.DataFrame(columns=list(df1_imp.columns) + ['File'])

for mz_rt in df1_imp['mz/RT']:
    column_name = str(mz_rt)

    if column_name in df2_wif.columns:
        max_concentration_row = df2_wif[column_name].idxmax()
        marker = df2_wif.iloc[max_concentration_row, 0]
        
        # Append the results to the new DataFrame
        result_df = result_df.append(pd.concat([df1_imp[df1_imp['mz/RT'] == mz_rt].iloc[0], pd.Series([marker], index=['File'])]), ignore_index=True)

        
# Write the result DataFrame to a new CSV file
result_df.to_csv('240702-NTA-Paper2-Unknown-Features-WIFFMatch.csv', index=False)
```
